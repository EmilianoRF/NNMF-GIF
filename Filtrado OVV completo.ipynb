```python
import os
import pandas as pd
import Funciones as Funciones
import scipy.signal as sig
```

```python
CARPETA     = "Señales VEC"
LISTA_TXT   = os.path.join(CARPETA, "Vocal I.txt")
COLUMNA_T   = "t_ms"
COLUMNA_OVV = "OVV"
```

```python
def leer_lista(path_txt):
    archivos = []
    with open(path_txt, "r", encoding="utf-8") as f:
        for linea in f:
            linea = linea.strip()
            if not linea or linea.startswith("Carpeta") or linea.startswith("Fecha") \
                    or linea.startswith("Total") or linea.startswith("-"):
                continue
            archivos.append(linea)
    return archivos
 
archivos = leer_lista(LISTA_TXT)
print(f"Archivos encontrados: {len(archivos)}")
```

```python
filtro_pb            = 'Chebyshev II'
#filtro_pb           = 'Butterworth'
orden_filtro_pb      = 10
atenuacion_filtro_pb = 40
fc_filtro_pb         = 500
orden_filtro_pa      = 6
fc_filtro_pa         = 60
metrica = 1
```

```python
for nombre in archivos:
    ruta = os.path.join(CARPETA, nombre)
    if not os.path.isfile(ruta):
        print(f"[OMITIDO] No se encontró: {ruta}")
        continue
 
    print(f"\nProcesando: {ruta}")
 
    # ── Leer el CSV ───────────────────────────
    df = pd.read_csv(ruta, sep=',')
 
    tiempos      = df[COLUMNA_T].values
    OVV_original = df[COLUMNA_OVV].values
 
    # ── PEGÁ TU CÓDIGO AQUÍ ───────────────────
    fs_r        = 8000
    OVV_original                   = OVV_original / max(abs(OVV_original))
    OVV_filtrada_pb, b_pb, a_pb    = Funciones.LP_Filter(OVV_original, fs=fs_r, fc=fc_filtro_pb, order=orden_filtro_pb)
    w_pb, h_pb                     = sig.freqz(b_pb, a_pb, fs=fs_r)
    OVV_filtrada_pb_pa, b_pa, a_pa = Funciones.HP_Filter(OVV_filtrada_pb, fs=fs_r)

    flujo_ovv, br, ar, FF, BW1 = Funciones.IF_SNF_OneMetric(OVV_filtrada_pb_pa, fs=fs_r, numMetric=metrica)

 
    # ── Guardar resultado en el mismo CSV ─────
    df['flujo_OVV'] = flujo_ovv
    df.to_csv(ruta, index=False, sep=',', float_format='%.6f')
    print(f"✓ Guardado: {ruta}")
 
print("\nProcesamiento finalizado.")
```
